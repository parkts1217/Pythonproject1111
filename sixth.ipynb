```python
def print_gugudan():
    """
    2단부터 9단까지 구구단을 출력하는 함수입니다.
    중첩 반복문을 사용하여 각 단의 계산 결과를 보여줍니다.
    """
    for dan in range(2, 10):  # 2단부터 9단까지 반복
        print(f"--- {dan}단 ---")
        for i in range(1, 10):  # 1부터 9까지 곱하기
            result = dan * i
            print(f"{dan} x {i} = {result}")
        print()  # 단 사이의 가독성을 위해 빈 줄 출력

# 함수 호출
print_gugudan()
```

```python
# 판다스에서 제공하는 기본 함수, house 데이터의 동별 거래가 평균 시간화
house = pd.read_csv('data\아산_아파트매매실거래_202604월.csv')
house['거래금액'] = house['거래금액'].str.replace(',', '').str.strip()
house['거래금액'] = pd.to_numeric(house['거래금액'], errors='coerce')   

house.groupby('법정동')['거래금액'].mean().plot(kind='bar', color='skyblue')

# matplotlib로 시각화
dong_data = house.groupby('법정동')['거래금액'].mean()
plt.bar(dong_data.index, dong_data.values)

# seaborn으로 시각화
sns.barplot(data=house, x='법정동', y='거래금액', estimator='mean')

plt.xlabel("법정동")
plt.ylabel("평균 거래금액")
plt.title("2026년 4월 아산 동별 거래가 평균 금액")
plt.xticks(rotation=45)
plt.show()
```
